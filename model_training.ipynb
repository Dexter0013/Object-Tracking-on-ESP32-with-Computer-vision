```python
# ====================== Install Ultralytics ======================
!pip install -q ultralytics

# ====================== Train YOLOv8 ======================
from ultralytics import YOLO

# Load model (start with nano)
model = YOLO("yolov8n.pt")

# Train
results = model.train(
    data="plam-Dataset-1/data.yaml",   # exact path
    epochs=100,
    imgsz=640,
    batch=16,                          # reduce to 8 if CUDA OOM
    patience=30,
    device=0,                          # GPU
    project="plam_yolo",
    name="yolov8n_plam",
    exist_ok=True,
    pretrained=True,
    optimizer="AdamW",
    lr0=0.001,
    plots=True,
    save=True,
    save_period=10
)

print("\nTraining finished!")
print(f"Best model saved at: {results.save_dir}/weights/best.pt")
```
